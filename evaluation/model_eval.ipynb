## Perplexity: https://huggingface.co/docs/transformers/perplexity



The core idea: perplexity asks "on average, how many equally likely choices did the model think it had at each token?" A PPL of 1 means the model was certain every time. A PPL of 10 means it felt like it was picking from 10 equally plausible options at each step. Lower is always better, but the number is only meaningful compared against the same vocabulary and dataset.
